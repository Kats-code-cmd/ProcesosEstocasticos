Suponiendo que tenemos el ejercicio siguiente:
"
En una celda LTE / 5G, los equipos de usuario  emplean el Random Access Channel (RACH) para solicitar conexión.

El estándar define 64 preámbulos ortogonales posibles (P = 64).

Suponga que, en un mismo “Random Access Opportunity”, N = 30 UEs intentan acceder simultáneamente y cada uno elige su preámbulo al azar con la misma probabilidad.

-¿Cuál es la probabilidad de que al menos dos UEs escojan el mismo preámbulo, provocando una colisión de acceso?
-Indique también el valor aproximado de N necesario para que esa probabilidad alcance el 50 %
-Explique el comportamiento de la probabilidad cuando aumenta el número de UEs. ¿Qué patrón observa?

Si tenemos un numero de UEs determinando como se comporta la probabilidad si modificamos la cantidad de preámbulos ortogonales posibles. ¿Qué patrón observa?

"

```python
#Calculo de 64 ^30 denominador
denominador= pow(64, 30)
```

Ahora, vamos a calcular el nominador del ejercicio el cual indica que:

```python

#Calculo del nominador
celdas= 64
numerador=1
for user in range(0,30):
 numerador=numerador*(celdas-user)
```

Denominador 

```python


#Calculo de las probabilidades
prob=numerador/denominador

```

```python
#probabilidad de colision
prob_col= 1-prob

print("Probabilidades de preámbulos ortogonales unicos ", prob)
print("Probabilidades de preámbulos ortogonales de colision ", prob_col)
```
